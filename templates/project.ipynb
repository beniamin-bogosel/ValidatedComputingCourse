# Project notebook template

**Authoring/student template.** Replace prompts as the project develops.

## Claim and scope

[What precisely will be proved? State domain, parameters, input uncertainty,
tolerance, and whether the claim is local or global.]

## Approximate exploration

[Show how numerical candidates were found. Mark heuristic choices clearly.]

## Mathematical validation

[State the theorem, all hypotheses, and how each is checked.]

## Implementation and certificate

[Provide executable code, proof-condition data, and a readable conclusion.
Record unresolved regions and budget limits.]

## Reproduction

[Software versions, precision, input encoding, seeds, and expected runtime.]

## Limitations and failure example

[What is not proved? Demonstrate a case that the method cannot certify.]

## Response to peer audit

[Record findings, corrections, and any changes in the final claim.]
